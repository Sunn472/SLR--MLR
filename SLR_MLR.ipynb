Q1. What is Simple Linear Regression?

Ans:-

Simple Linear Regression is a statistical method used to model the relationship between one independent variable (predictor) and one dependent variable (response) by fitting a straight line to the observed data. The goal is to predict the dependent variable based on the independent variable.

Dependent Variable (Y): The outcome you want to predict.

Independent Variable (X): The predictor used to explain changes in Y.

Regression Line: The best-fit straight line defined by the equation:

Y=β0+β1X

β0=intercept(value of y when x=0)

β1=slope(change in y wrt xhange in x)

Q2.  What are the key assumptions of Simple Linear Regression?

Ans:-

For valid results, the model assumes:

Linearity: The relationship between X and Y is linear.

Independence: Residuals are independent (no autocorrelation).

Homoscedasticity: Residuals have constant variance across X.

Normality: Residuals are approximately normally distributed (important for inference)

Q3. What does the coefficient m represent in the equation Y=mX+c?

Ans:-

The coefficient m (slope) represents:

"The change in the dependent variable (Y) for a one-unit increase in the independent variable (X)."

Q4. What does the intercept c represent in the equation Y=mX+c?

Ans:-

Interpretation of the Intercept (c)

Mathematical Meaning:

It's the point where the regression line crosses the Y-axis (i.e., when X=0, Y=c).

Practical Meaning:

If X=0 is within the observed data range, c gives the expected baseline value of Y.

If X=0 is outside the data range, c may not have a meaningful interpretation (extrapolation risk).

Q5. How do we calculate the slope m in Simple Linear Regression?

Ans:-

Use the Least Squares Method:

m=∑(Xi−Xˉ)(Yi−Yˉ)/∑(Xi−Xˉ)^2

Steps:
1. Compute means Xˉ and Yˉ.
2. Calculate covariance (numerator) and variance of X(denominator).
3. Divide covariance by variance

Q6. What is the purpose of the least squares method in Simple Linear Regression?

Ans:-

Purpose of Least Squares Method is to Minimizes the sum of squared residuals (errors) between observed and predicted Y,ensuring the best-fit line.

Q7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?

Ans:-

1. Definition: Proportion of variance in Y explained by X.
2. Range: 0 (no explanation) to 1 (perfect explanation).
3. Limitation: Doesn’t indicate causation or model correctness

Q10. What are the key assumptions of Multiple Linear Regression?

Ans:-

Assumptions of MLR(Multiple Linear Regression):-
1. Linearity: The relationship between X and Y is linear.
2. Independence: Residuals are independent (no autocorrelation).

Q11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Heteroscedasticity
1. Definition: Non-constant variance in residuals.
2. Effect: Biases standard errors, affecting hypothesis tests.
3. Fix: Transform Y (e.g., log) or use robust standard errors.

Q23. What is polynomial regression?

Ans:-

Polynomial Regression is a form of nonlinear regression that models the relationship between the independent variable (X) and the dependent variable (Y) as an n-th degree polynomial. It extends Simple Linear Regression by adding polynomial terms (X2,X2,X3,…) to capture curvature in the data

Q24. How does polynomial regression differ from linear regression?

Ans:-

Polynomial regression extends linear regression by adding polynomial terms (like X^2,X^3) to model nonlinear relationships, while linear regression assumes a straight-line relationship (Y=mX+c). Polynomial regression fits curves and can capture complex patterns, but risks overfitting with high degrees. Linear regression is simpler and works only for linear trends. Both use least squares to fit the model, but polynomial regression offers flexibility at the cost of complexity

Q25. When is polynomial regression used?

Ans:-

Polynomial regression is used when:
1. The relationship between variables is nonlinear (e.g., curved trends like growth rates, acceleration, or diminishing returns).
2. Linear regression underfits the data, failing to capture patterns visible in scatterplots.
3. Higher accuracy is needed for complex trends without switching to more advanced models (e.g., neural networks).
3. Domain knowledge suggests a polynomial relationship (e.g., physics laws like projectile motion).

Q26. What is the general equation for polynomial regression?

Q27. Can polynomial regression be applied to multiple variables?

Ans:-

Yes, polynomial regression works with multiple variables by adding polynomial and interaction terms. It’s powerful but requires careful feature engineering to avoid overfitting. Use it when:
1. Relationships are nonlinear and involve synergies between predictors.
2. Domain knowledge justifies polynomial terms.
3. Dataset size supports the added complexity.

Q28. What are the limitations of polynomial regression?

Ans:-

Here are the key limitations of polynomial regression in concise points:
1. Overfitting- High-degree polynomials fit noise instead of the true trend, especially with small datasets.
2. Computationally Expensive- More features (like X^2,X^3,X1,X2) increase model complexity and training time.
3. Poor Extrapolation- Predictions outside the training data range are often unreliable.
4. Sensitive to Outliers- Polynomial curves can bend drastically to fit extreme values.
5. Multicollinearity- Higher-order terms (X^2,X^3)correlate with original features, skewing coefficients.
6. Interpretability- Complex models with interaction terms (X1,X2)are harder to explain.
7. Curse of Dimensionality- Adding many variables/powers leads to sparse data in high-dimensional space.
8. Risk of Underfitting- Low-degree polynomials may miss true nonlinear patterns.

Q29.  What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Ans:-

To evaluate polynomial regression fit and select the optimal degree, use these key methods:
1. Cross-Validation (MSE/R²) - Most reliable; tests performance on unseen data (e.g., sklearn.cross_val_score).
2. Adjusted R² - Penalizes unnecessary complexity; prefer higher values.
3. AIC/BIC - Balance fit and parsimony; lower values = better.
4. Train-Test Split - Compare training vs. test MSE to detect overfitting.
5. Visualization - Plot curves to spot overfitting (wiggles) or underfitting (oversimplified).

Q30. Why is visualization important in polynomial regression?

Ans:-

Importantance of visualization  in Polynomial Regression:-
1. Identifies Nonlinear Patterns- Helps spot curves, bends, or complex trends that justify polynomial terms.
2. Detects Overfitting/Underfitting- Visual checks reveal if the model fits noise (overfitting) or misses trends (underfitting).
3. Validates Extrapolation- Shows if predictions behave erratically outside the data range.
4. Assesses Residuals- Residual plots (e.g., random vs. patterned) indicate model adequacy.
5. Compares Degrees- Side-by-side plots of different polynomial degrees guide optimal selection.
6. Reveals Outliers- Highlights data points that distort the polynomial curve.
7. Supports Interpretation- Makes abstract coefficients (e.g. X1,X2)tangible through observed curves.

```python
'''
Q31. How is polynomial regression implemented in Python?
'''
#import libararies
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#preapre data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Feature
y = np.array([2, 4, 5, 7, 10])                # Target

#Transform Features to Polynomial Terms
degree = 2  # Quadratic fit
poly = PolynomialFeatures(degree=degree, include_bias=False)
X_poly = poly.fit_transform(X)  # Converts X to [X, X², ...]

#Fit the Model
model = LinearRegression()
model.fit(X_poly, y)

#Predict & Evaluate
y_pred = model.predict(X_poly)
print("Coefficients:", model.coef_)  # Slopes for X, X², ...
print("Intercept:", model.intercept_)
print("R² Score:", r2_score(y, y_pred))

#Visualize the fit
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', label=f'Degree {degree} Fit')
plt.legend()
plt.show()
```
